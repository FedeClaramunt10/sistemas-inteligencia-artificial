## Descargar y Limpiar el Conjunto de Datos


```python
filepath='https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/DL0101EN/labs/data/concrete_data.csv'
concrete_data = pd.read_csv(filepath)

concrete_data.head()
```

```python
concrete_data.describe()
```

```python
concrete_data.isnull().sum()
```

Los datos parecen muy limpios y están listos para ser utilizados en la construcción de nuestro modelo.


#### Dividir los datos en predictores y objetivo


```python
concrete_data_columns = concrete_data.columns
concrete_data_columns
```

```python
predictors = concrete_data[concrete_data_columns[concrete_data_columns != 'Strength']] # all columns except Strength
target = concrete_data['Strength'] # Strength column
```

```python
predictors.head()
```

```python
target.head()
```

Finalmente, el último paso es normalizar los datos restando la media y dividiendo por la desviación estándar.


```python
predictors_norm = (predictors - predictors.mean()) / predictors.std()
predictors_norm.head()
```

```python
n_cols = predictors_norm.shape[1] # cantidad de predictores
n_cols
```

## Importar Paquetes de Keras

##### Vamos a importar el resto de los paquetes de la biblioteca Keras que necesitaremos para construir nuestro modelo de regresión.


```python
from keras.models import Sequential
from keras.layers import Dense
from keras.layers import Input
```

## Construir una Red Neuronal


Definamos una función que defina nuestro modelo de regresión para que podamos llamarla convenientemente para crear nuestro modelo.


```python
# define regression model
def regression_model():
    # create model
    model = Sequential()
    model.add(Input(shape=(n_cols,)))
    model.add(Dense(50, activation='relu'))
    model.add(Dense(50, activation='relu'))
    model.add(Dense(1))

    # compile model
    model.compile(optimizer='adam', loss='mean_squared_error')
    return model
```

## Entrenar y Probar la Red


```python
# contruimos el modelo
model = regression_model()
```

```python
# ajustamos o entrenamos el modelo
# Entrenamiento con validación
history = model.fit(
    predictors_norm,
    target,
    validation_split=0.3,   # 30% de los datos se usan para validación
    epochs=25,
    verbose=2
)
```

```python
# Graficar las curvas de pérdida
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
plt.plot(history.history['loss'], label='Entrenamiento')
plt.plot(history.history['val_loss'], label='Validación')
plt.xlabel('Épocas')
plt.ylabel('Pérdida (MSE)')
plt.title('Curvas de pérdida - Evaluación de sobreajuste')
plt.legend()
plt.show()
```

<h3>Ejercicio de Práctica 1</h3>


Ahora, usando el mismo conjunto de datos, intenta recrear un modelo de regresión que cuente con cinco capas ocultas, cada una con 50 nodos y funciones de activación ReLU, y una única capa de salida, optimizado utilizando el optimizador Adam.


```python
# SOLUCIÓN EJERCICIO 1: Modelo con 5 capas ocultas
def regression_model_5capas():
    model = Sequential()
    model.add(Input(shape=(n_cols,)))
    model.add(Dense(50, activation='relu'))
    model.add(Dense(50, activation='relu'))
    model.add(Dense(50, activation='relu'))
    model.add(Dense(50, activation='relu'))
    model.add(Dense(50, activation='relu'))
    model.add(Dense(1))

    model.compile(optimizer='adam', loss='mean_squared_error')
    return model
```

<h3>Ejercicio Práctico 2</h3>


Entrena y evalúa el modelo simultáneamente utilizando el método fit() reservando el 10% de los datos para validación y entrenando el modelo durante 100 épocas.


```python
# SOLUCIÓN EJERCICIO 2: Entrenar con 10% validación y 100 épocas
modelo = regression_model()
history_m1 = modelo.fit(predictors_norm, target, validation_split=0.1, epochs=100, verbose=2)
```

```python
# Graficar las curvas de pérdida
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
plt.plot(history_m1.history['loss'], label='Entrenamiento')
plt.plot(history_m1.history['val_loss'], label='Validación')
plt.xlabel('Épocas')
plt.ylabel('Pérdida (MSE)')
plt.title('Curvas de pérdida - Modelo con 10% validación y 100 épocas')
plt.legend()
plt.show()
```

## Conclusiones

- **Más capas ocultas**: Agregar más capas ocultas al modelo aumenta su capacidad para aprender y representar relaciones complejas dentro de los datos.
